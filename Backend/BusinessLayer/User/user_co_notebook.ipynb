```python
import re
def is_valid_password(password):
    """
    Validate the password.
    Password must:
    - Be at least 8 characters long
    - Be at most 20 characters long
    - Contain only English letters (a-z, A-Z), numbers, and allowed special characters
    - Contain at least one uppercase letter
    - Contain at least one lowercase letter
    - Contain at least one number
    - Contain at least one special character: {, }, [, ], !, @, $, %, ^, &, *, (, ), +
    """

    # Check for minimum and maximum length
    if len(password) < 8 or len(password) > 20:
        return False

    # Ensure password contains only allowed characters
    if not re.match(r"^[A-Za-z0-9{}\[\]!@\$%\^&\*\(\)\+]+$", password):
        return False

    # Regular expressions for each condition
    has_uppercase = re.search(r"[A-Z]", password)
    has_lowercase = re.search(r"[a-z]", password)
    has_number = re.search(r"[0-9]", password)
    has_special = re.search(r"[{}\[\]!@\$%\^&\*\(\)\+]", password)

    # Return True only if all conditions are met
    return bool(has_uppercase and has_lowercase and has_number and has_special)
```

```python
print(is_valid_password("Password1!"))   # True (valid)
print(is_valid_password("Pass!"))        # False (too short)
print(is_valid_password("Password1!א"))  # False (contains non-English letter)
print(is_valid_password("P@ssword1234567890123"))  # False (too long)
print(is_valid_password("password123!")) # False (no uppercase letter)
print(is_valid_password("PASSWORD123!")) # False (no lowercase letter)
print(is_valid_password("Password!"))    # False (no number)
```

```python

def is_valid_name(name):
    """
    Validates a name to ensure it contains only Hebrew characters, spaces, tabs, and hyphens.

    Args:
    name (str): The name to validate.

    Returns:
    bool: True if the name is valid, False otherwise.
    """
    # Regular expression to allow only Hebrew characters, spaces, tabs, and hyphens
    hebrew_name_regex = r'^[\u0590-\u05FF]+([\s\t-][\u0590-\u05FF]+)*$'
    
    # Validate the name using regex
    return bool(re.match(hebrew_name_regex, name))
```

```python
# Examples of usage:
print(is_valid_name("דוד כהן"))       # True
print(is_valid_name("דוד-כהן"))       # True
print(is_valid_name("דוד\tכהן"))      # True
print(is_valid_name("דוד כהן-לוי"))   # True
print(is_valid_name("David Cohen"))   # False
print(is_valid_name("דוד123 כהן"))    # False
print(is_valid_name("דוד כהן@"))      # False
print(is_valid_name("ככ"))       # True
print(is_valid_name("וולודרסקי"))       # True

```

```python
from PyPDF2 import PdfReader
import re

def extract_syllabus_topics(file_path):
    """
    Extracts syllabus topics from a PDF file, adapting to various formats like headers, free text, and tables.

    :param file_path: Path to the PDF file
    :return: A set of extracted syllabus topics
    """
    reader = PdfReader(file_path)
    syllabus_topics = set()

    # Define potential headers for syllabus sections
    headers_patterns = [
        r'סילבוס באנגלית[:\n](.*?)\n',
        r'סילבוס בעברית[:\n](.*?)\n',
        r'נושא השיעור[:\n](.*?)\n',
        r'נושאי השיעור[:\n](.*?)\n',
        r'Course Topics[:\n](.*?)\n',
        r'Outline[:\n](.*?)\n'
    ]

    # Patterns for recognizing table-like structures
    table_header_patterns = [
        r'נושא השיעור',
        r'נושאי השיעור',
        r'Topic',
        r'Topics'
    ]

    # Combine all text from the PDF for global search
    pdf_text = "".join(page.extract_text() for page in reader.pages)

    # Handle specific headers and free text sections
    for pattern in headers_patterns:
        matches = re.findall(pattern, pdf_text, re.DOTALL)
        for match in matches:
            topics = re.split(r"\n|\u2022|,|;", match)  # Split by common delimiters
            syllabus_topics.update(topic.strip() for topic in topics if topic.strip())

    # Handle table-like structures
    for page in reader.pages:
        lines = page.extract_text().split("\n")
        for i, line in enumerate(lines):
            if any(header in line for header in table_header_patterns):
                # Extract rows under the identified table header
                for row in lines[i + 1:]:
                    if row.strip() == "" or re.match(r"^\s*-+\s*$", row):  # End of table
                        break
                    syllabus_topics.add(row.strip())

    return syllabus_topics
```

```python
from PyPDF2 import PdfReader
import re


def extract_syllabus_topics3(file_path):
    """
    Extracts syllabus topics from a course PDF file and returns them as a dictionary.
    High-priority sections are processed first, and if found, lower-priority patterns are skipped.

    :param file_path: Path to the PDF file
    :return: A dictionary with keys as section headers and values as sets of topics
    """
    reader = PdfReader(file_path)
    syllabus_results = {}

    # Higher-priority patterns
    high_priority_patterns = {
        r'סילבוס באנגלית[:\n](.*?)\n': "סילבוס באנגלית",
        r'סילבוס בעברית[:\n](.*?)\n': "סילבוס בעברית",
        r'סילבוס[:\n](.*?)\n': "סילבוס",
    }

    # Lower-priority patterns
    lower_priority_patterns = [
        r'נושאים[:\n](.*?)\n',  # Hebrew pattern for "Topics"
        r'Course Topics[:\n](.*?)\n',
        r'Outline[:\n](.*?)\n',
    ]

    def extract_topics_from_text(text, patterns):
        """
        Helper function to extract topics based on patterns.
        """
        topics = set()
        for pattern in patterns:
            matches = re.findall(pattern, text, re.DOTALL)
            for match in matches:
                # Split potential topics by common delimiters and clean up
                topics.update([topic.strip() for topic in re.split(r',|;|\n|\•|•', match) if topic.strip()])
        return topics

    for page in reader.pages:
        text = page.extract_text()

        # Process high-priority patterns
        for pattern, header in high_priority_patterns.items():
            matches = re.findall(pattern, text, re.DOTALL)
            if matches:
                if header not in syllabus_results:
                    syllabus_results[header] = set()
                for match in matches:
                    # Extract and clean topics
                    topics = [topic.strip() for topic in re.split(r'\n|\•|•', match) if topic.strip()]
                    syllabus_results[header].update(topics)

        # If high-priority patterns were found, skip lower-priority patterns
        if syllabus_results:
            continue

        # Process lower-priority patterns
        lower_topics = extract_topics_from_text(text, lower_priority_patterns)
        if lower_topics:
            if "Other Topics" not in syllabus_results:
                syllabus_results["Other Topics"] = set()
            syllabus_results["Other Topics"].update(lower_topics)

    return syllabus_results

# Example usage
file_path = "/Users/davidvolodarsky/Desktop/Semeters/Semester_G/NegevNerds/sylbus_analyzer/data_strcutre.pdf"  # Replace with the correct path to the PDF
file_pdf1 = "/Users/davidvolodarsky/Desktop/Semeters/Semester_G/NegevNerds/sylbus_analyzer/bsisi_netunim.pdf"
result = extract_syllabus_topics3(file_path)
# print("Extracted Topics:", syllabus_topics)

# Print the results
for header, topics in result.items():
    print(f"Section: {header}")
    print("Topics:")
    for topic in topics:
        print(f"- {topic}")
```

```python
from PyPDF2 import PdfReader
import re

def extract_syllabus_topics4(file_path, topic_patterns):
    """
    Extracts syllabus topics from a course PDF file and returns them as a set.
    Handles diverse formats such as tables, bullet points, numbered sections, and headers.

    :param file_path: Path to the PDF file
    :param topic_patterns: List of regex patterns to identify syllabus-related sections
    :return: A set of topics from the syllabus
    """
    reader = PdfReader(file_path)
    syllabus_topics = set()

    for page in reader.pages:
        text = page.extract_text()

        # Match topics using provided patterns
        for pattern in topic_patterns:
            matches = re.findall(pattern, text, re.DOTALL)
            for match in matches:
                # Split potential topics by common delimiters and clean up
                topics = re.split(r',|;|\n|\•|\.', match)
                syllabus_topics.update([topic.strip() for topic in topics if topic.strip()])

        # Handle bullet points
        lines = text.split("\n")
        for line in lines:
            if re.match(r'^\•', line):  # Matches lines starting with "•"
                syllabus_topics.add(line.lstrip("• ").strip())

        # Handle numbered sections (e.g., "1. Topic", "2. Topic")
        for line in lines:
            if re.match(r'^\d+\.\s', line):  # Matches lines starting with "1. ", "2. ", etc.
                syllabus_topics.add(line.strip())

        # Handle keywords directly in the text
        if any(keyword in text for keyword in ["סילבוס", "Topics", "Outline"]):
            for line in lines:
                # Add lines containing relevant keywords as potential topics
                if any(keyword in line for keyword in ["סילבוס", "Topics", "Outline"]):
                    syllabus_topics.add(line.strip())

    return syllabus_topics



# def extract_syllabus_topics4(file_path, topic_patterns):
#     """
#     Extracts syllabus topics from a course PDF file and returns them as a set.
#     Handles diverse formats such as tables, bullet points, and headers.

#     :param file_path: Path to the PDF file
#     :param topic_patterns: List of regex patterns to identify syllabus-related sections
#     :return: A set of topics from the syllabus
#     """
#     reader = PdfReader(file_path)
#     syllabus_topics = set()

#     for page in reader.pages:
#         text = page.extract_text()
#         for pattern in topic_patterns:
#             matches = re.findall(pattern, text, re.DOTALL)
#             for match in matches:
#                 # Ensure the match contains both words in the pattern if it has two words
#                 if len(pattern.split()) == 2:  # Check if the pattern has two words
#                     words = pattern.split()
#                     if all(word in match for word in words):
#                         # Split potential topics by common delimiters and clean up
#                         topics = re.split(r',|;|\n|\•|\•', match)
#                         syllabus_topics.update([topic.strip() for topic in topics if topic.strip()])
#                 else:
#                     # Split potential topics by common delimiters and clean up
#                     topics = re.split(r',|;|\n|\•|\•', match)
#                     syllabus_topics.update([topic.strip() for topic in topics if topic.strip()])

#         # Handle topics found in tables or other structured formats
#         if "סילבוס" in text or "נושאי השיעור" in text or "Topics" in text or "Outline" in text:
#             lines = text.split("\n")
#             for line in lines:
#                 if re.search(r'\•|\•', line):  # Look for bullet points
#                     syllabus_topics.add(line.strip())

#     return syllabus_topics

# def extract_syllabus_topics4(file_path, topic_patterns):
#     """
#     Extracts syllabus topics from a course PDF file and returns them as a set.
#     Handles diverse formats such as tables, bullet points, and headers.

#     :param file_path: Path to the PDF file
#     :param topic_patterns: List of regex patterns to identify syllabus-related sections
#     :return: A set of topics from the syllabus
#     """
#     reader = PdfReader(file_path)
#     syllabus_topics = set()

#     for page in reader.pages:
#         text = page.extract_text()
#         for pattern in topic_patterns:
#             matches = re.findall(pattern, text, re.DOTALL)
#             for match in matches:
#                 # Split potential topics by common delimiters and clean up
#                 topics = re.split(r',|;|\n|\•|\•', match)
#                 syllabus_topics.update([topic.strip() for topic in topics if topic.strip()])

#         # Handle topics found in tables or other structured formats
#         if "סילבוס" in text or "נושאי השיעור" or "Topics" in text or "Outline" in text:
#             lines = text.split("\n")
#             for line in lines:
#                 if re.search(r'\•|\•', line):  # Look for bullet points
#                     syllabus_topics.add(line.strip())

#     return syllabus_topics

# # Example usage
# file_path = "/Users/davidvolodarsky/Desktop/Semeters/Semester_G/NegevNerds/sylbus_analyzer/data_strcutre.pdf"  # Replace with the correct path to the PDF
# file_pdf1 = "/Users/davidvolodarsky/Desktop/Semeters/Semester_G/NegevNerds/sylbus_analyzer/bsisi_netunim.pdf"

# # Define topic patterns externally
# topic_patterns = [
#     r'סילבוס[:\n](.*?)\n',  # Hebrew pattern for "Syllabus"
#     r'סילבוס באנגלית[:\n](.*?)\n',
#     r'סילבוס בעברית[:\n](.*?)\n',
#     r'נושאים[:\n](.*?)\n',  # Hebrew pattern for "Topics"
#     r'Course Topics[:\n](.*?)\n',
#     r'Outline[:\n](.*?)\n',
# ]

# syllabus_topics = extract_syllabus_topics1(file_path, topic_patterns)
# print("Extracted Topics:", syllabus_topics)
# print(len(syllabus_topics))
```

```python
from PyPDF2 import PdfReader
import re

def extract_topics_from_table(file_path, topic_patterns):
    """
    Extracts syllabus topics from a course PDF file, handling diverse formats like tables and bullet points.
    Only extracts data from the relevant column in tables.

    :param file_path: Path to the PDF file
    :param topic_patterns: List of regex patterns to identify syllabus-related sections
    :return: A set of topics from the syllabus
    """
    reader = PdfReader(file_path)
    syllabus_topics = set()

    for page_num, page in enumerate(reader.pages):
        text = page.extract_text()
        
        # Handle table extraction for columns matching topic_patterns
        lines = text.split("\n")
        capture_table = False
        header_index = -1

        for line in lines:
            # Check if the line matches any of the topic_patterns
            for pattern in topic_patterns:
                if re.search(pattern, line):  # Locate the header row
                    capture_table = True
                    headers = line.split()  # Split columns in the header row
                    for i, header in enumerate(headers):
                        if re.search(pattern, header):  # Check if header matches the pattern
                            header_index = i  # Identify the relevant column index
                            break
                    break  # Stop checking other patterns if a match is found

            if capture_table and header_index != -1:
                # Extract topics from subsequent rows
                columns = line.split()  # Split row into columns
                if len(columns) > header_index:
                    # Extract only the topic from the relevant column
                    topic = columns[header_index].strip()
                    if topic:  # Add the topic if it's not empty
                        syllabus_topics.add(topic)

        # Handle topics matching the predefined patterns in non-table format
        for pattern in topic_patterns:
            matches = re.findall(pattern, text, re.DOTALL)
            for match in matches:
                # Split potential topics by common delimiters and clean up
                topics = re.split(r',|;|\n|\•|•', match)
                syllabus_topics.update([topic.strip() for topic in topics if topic.strip()])

    return syllabus_topics


```
